```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('https://raw.githubusercontent.com/biofreelancer/public_datasets/refs/heads/main/cursos_especiales/edicion_graficos_cientificos/ejercicio_2/remoto.tsv', sep='\t')

df.head( )
```

```python
# Iniciar el plot
fig, ax = plt.subplots( figsize=( 7, 7 ) )  # 7 x 7 pulgadas, ancho por alto

ax.plot(
    [1, 1], [0, 1],
    transform = ax.get_xaxis_transform( ),  # extender la linea negra en x = 1
    clip_on = False
)

# Dibujar cada factor como circulo y bigotes
for i in range(len(df)):
    
    ax.errorbar(
        df['OR'][i], i, 
        xerr = [[df['err_lower'][i]], [df['err_upper'][i]]], 
        fmt = 'o', # poner un punto solido
        markersize = 10, # tamanio del punto central
        clip_on = False
    )

# agregar nombres eje Y
ax.set_yticks( range( len( df ) ) )
ax.set_yticklabels( df['Factor'] )

# Guardar en PDF
plt.savefig( 'forest_ok.pdf', bbox_inches='tight' )

# Bajarlo en PDF del menu de la izquierda
```
